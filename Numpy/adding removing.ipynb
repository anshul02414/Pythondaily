Add elements

```python
import numpy as np

a = np.array([1,2,3,4])

np.append(a,90)
```

```python
a = np.array([[1,2],[3,4]]) # 2d array convert into 1d arrey the append aperation perform

np.append(a, 400)
```

```python
np.append(a, [400, 100, 50])
```

insert

```python
np.insert(a, [1], [20,50], axis=1) # array index value
```

Delete

```python
print(a)
np.delete(a,0,axis=0)  # convert arrey into 1d then delete index

# axis=0  -> delete the row
```

```python
print(a)
np.delete(a,0,axis=1) 

# axis=1  -> delete the column
```
