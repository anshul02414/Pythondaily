Boolean Masking

```python
arr = np.array([12,45,7,89,34,22,100])
```

Print all elements greater than 30.

```python
fa = arr > 30
newarr = arr[fa]
print(newarr)
```

Print all even numbers.

```python
fa = arr % 2 == 0
newarr = arr[fa]
print(newarr)
```

Count how many numbers are greater than 50.


```python
fa = arr > 50
mewarr = arr[fa]
print(np.size(newarr))
```

Replace all numbers less than 20 with 0.


```python
fa = arr < 20
arr[fa] = 0

print(arr)
```

Find all prime numbers between 1 and 100 using NumPy.


```python
arr = np.arange(1,101)

for i in arr:
    if i < 2:
        continue
    
    isPrime = True
    
    for j in range(2, i):
        if i % j == 0:
            isPrime = False
            break
    
    if isPrime:
        print(i, end=" ")
```
