Join

```python
# join 2 arrey -> concatenate

import numpy as np

arr1 = np.array([[1,2],
                 [3,5]])

arr2 = np.array([[4,5],
                 [6,9]])

print(arr1)
print(arr2)

print(np.concatenate([arr1, arr2]))
```

```python
print(np.concatenate([arr1, arr2], axis=1))
```

```python
print(np.hstack([arr1, arr2]))
```

```python
print(np.vstack([arr1, arr2]))
```
