Aggrating fn

```python
import numpy as np

l = [7,8,2,4,9,6,4,1]

arr = np.array(l)
print(arr)
```

```python
print(np.sum(arr))
```

```python
print(np.min(arr))
print(np.max(arr))
print(np.size(arr))
```

```python
print(np.mean(arr))
```

```python
print(np.cumsum(arr))
print(np.cumprod(arr))
```

```python
a = [100, 150, 199, 200, 250, 130]
b = [10, 50, 30, 40, 30, 10]

price = np.array(a)
quantity = np.array(b)

print(price)
print(quantity)
```

```python
c = np.cumprod([price, quantity], axis=0)
print(c)
```

```python
print(c[1].sum())
```

statical function

```python
l = [7,8,2,4,9,6,4,1]

arr = np.array(l)

print(arr)
```

```python
import statistics as state

print(state.mode(arr))
```

```python
print(np.std(arr))
```

```python
print(np.var(arr))
```

```python
a = [100, 150, 199, 200, 250, 130]
b = [10, 50, 30, 40, 30, 10]

price = np.array(a)
quantity = np.array(b)

print(np.corrcoef([price, quantity]))  # shows how 2 arrays are relate with each other


"""
-1 = Inversaly proportional
0 = No relation
1 = Proportional
"""
```
