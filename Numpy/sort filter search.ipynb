Sort

```python
import numpy as np

arr = np.array([8,7,6,1,74,8,6])

print(np.sort(arr))
```

```python
print(np.sort(arr)[::-1])
```

Search

```python
arr = np.array([8,7,6,1,74,8,6])
```

```python
s = np.where(arr == 99)  # return index of the values
print(s)

s = np.where(arr == 74)
print(s)
```

```python
s = np.where(arr > 5)  # return index of the values
print(s)
```

```python
s = np.where(arr%2 == 0)  # return index of the values
print(s)
```
