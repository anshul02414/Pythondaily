```python
import pandas as pd
```

```python
d = {"idx":[1,2,3,4,5],
     "Name":["Anshul","Raj","Ram","Shyam","Koi"],
     "Age":[22,24,19,42,33]}

data = pd.DataFrame(d)
data
```

```python
print(data.pivot(index="idx", columns="Name", values="Age"))
```

```python
data["city"] = ["jaipur","Delhi","Jaipur","Gujrat","Delhi"]
```

```python
data
```

```python
print(data.pivot(index="idx", columns="Name", values=["Age","city"]))
```

Melting

```python
data = {"Name":["Anshul","Raj","Priya","Shyam","Koi"],
        "Gender":["M","M","F","M","F"],
        "Age":[22,24,19,42,33]}

df = pd.DataFrame(data)
```

```python
df
```

```python
print(pd.melt(df, id_vars=["Name"], value_vars=["Gender"]))
```

```python
print(pd.melt(df, id_vars=["Name"], value_vars=["Gender", "Age"]))
```

```python
print(pd.melt(df, id_vars=["Name"], value_vars=["Gender"], var_name="Var_gender"))
```
