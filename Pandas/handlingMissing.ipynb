```python
import pandas as pd

data = pd.read_csv(
    "employ_Data.csv",
    engine="python")
data.head(5)
```

```python
# handling missing values

print(data.isnull())
```

```python
print(data.isnull().sum())
```

```python
import numpy as np
```

fill name

```python
data["Name"] = data["Name"].fillna("Noname")
```

```python
print(data.isnull().sum())
```

fill department

```python
data["Department"] = data["Department"].fillna(method="bfill")
```

```python
print(data.isnull().sum())
```

```python
print(data)
```

salary filling

```python
data["Salary"] = pd.to_numeric(data["Salary"], errors="coerce")
```

```python
meanval = data["Salary"].mean()
print(meanval)
```

```python
data["Salary"] = data["Salary"].fillna(meanval)
```

```python
print(data.isnull().sum())
```

age

```python
data["Age"] = data["Age"].fillna(method="bfill")
```

```python
print(data.isnull().sum())
```

Joining Date

```python
data["Joining_Date"] = data["Joining_Date"].fillna(method="ffill")
```

```python
print(data.isnull().sum())
```

city

```python
data["City"] = data["City"].fillna(method="ffill")
```

```python
print(data.isnull().sum())
```

Performance

```python
data["Performance_Score"] = pd.to_numeric(data["Performance_Score"], errors="coerce")
```

```python
meanp = data["Performance_Score"].mean()
```

```python
avgP = data["Performance_Score"].mean()
data["Performance_Score"] = data["Performance_Score"].fillna(avgP)
```

```python
print(data.isnull().sum())
```

Email

```python
data["email"] = data["email"].fillna(
    data["Name"].str.lower() + "gmail.com"
)
```

```python
print(data.isnull().sum())
```

```python
print(data[data["email"].isna()])
```
