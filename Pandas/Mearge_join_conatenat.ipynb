```python
data1 = {"EmpId":["E01","E02","E03","E04","E05","E06"],
         "Name":["Anshul","Ram","Shyam","Raju","Koi","Rajesh"],
         "Age":[33,25,45,21,25,27]}

data2 = {"EmpId":["E01","E07","E03","E04","E05","E08"],
         "Salary":[45000,56000,70000,19000,24000,29000]}

df1 = pd.DataFrame(data1)
df2 = pd.DataFrame(data2)
```

```python
df1
```

```python
df2
```

```python
pd.merge(df1,df2,on="EmpId", how="left")
```

```python
pd.merge(df1,df2,on="EmpId", how="right")
```

```python
pd.merge(df1,df2,on="EmpId",how="outer")
```

```python
data1 = {"EmpId":["E01","E02","E03","E04","E05","E06"],
         "Name":["Anshul","Ram","Shyam","Raju","Koi","Rajesh"],
         "Age":[33,25,45,21,25,27]}

data2 = {"EmployId":["E01","E07","E03","E04","E05","E08"],
         "Salary":[45000,56000,70000,19000,24000,29000]}
```

```python
df1 = pd.DataFrame(data1)
df2 = pd.DataFrame(data2)
```

```python
pd.merge(df1, df2, left_on="EmpId", right_on="EmployId")
```

```python
pd.merge(df1, df2, left_on="EmpId", right_on="EmployId",indicator=True)
```

Concatenate

```python
data1 = {"EmpId":["E01","E02","E03","E04","E05","E06"],
         "Name":["Anshul","Ram","Shyam","Raju","Koi","Rajesh"],
         "Age":[33,25,45,21,25,27]}

data2 = {"EmpId":["E07","E08","E09","E10","E011","E012"],
         "Name":["Bittu","Kittu","Chittu","Ritu","ramshyam","Ghanshyam"],
         "Age":[45,56,27,19,24,29]}

df1 = pd.DataFrame(data1)
df2 = pd.DataFrame(data2)
```

```python
print(pd.concat([df1,df2]))
```
