```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_excel("add row practice.xlsx")
data.head(5)
```

```python
# Salary ka 20% Bonus column banao.
data["Bonas"] = (data["Salary"]/100)*25
```

```python
data.head(5)
```

```python
# Salary + Bonus = Total Salary column banao.
data["Total Salary"] = data["Salary"] + data["Bonas"]
```

```python
data.head(4)
```

```python
# Name se email column banao.
data["Email"] = data["Name"].str.lower() + "@gmail.com"
```

```python
data.head(5)
```

```python
# Age > 30 ho to "Senior" warna "Junior".
data["Post"] = np.where(
    data["Age"] > 30,
    "Senior",
    "Jonior"
)
```

```python
data.head(5)
```

```python
# Salary ko lakh me convert karo.
data["Salary in Lacs"] = data["Total Salary"]/100000
```

```python
data.head(5)
```

```python
# Joining_Date se Year aur Month nikalo.
data["Joinong_year"] = data["Joining Date"].dt.year
```

```python
data.head(3)
```

```python
data["Joining_Month"] = data["Joining Date"].dt.month
```

```python
data.head(4)
```

```python
data["Joining_Month name"] = data["Joining Date"].dt.month_name()
```

```python
data.head(3)
```

```python
# Salary ranking column banao.
data["ranking"] = data["Salary"].rank(ascending=False)
```

```python
data.head(4)
```

```python
# Running total of salary nikalo.
data["Running Salary"] = data["Salary"].cumsum()
```

```python
data.head(4)
```

```python
# Salary ka percentage contribution nikalo.
t = data["Salary"].sum()
data["Contribution"] = (data["Salary"]/t)*100
```

```python
data.head(5)
```

```python
# Name ke first 3 letters wala column banao.
data["sort Name"] = data["Name"].str[0:3]

# data["sort Name"] = data["Name"].str.slice(0,3)
```
