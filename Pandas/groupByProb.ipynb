```python
# Har employee ke liye department average salary column banao (transform).

data["Avg_salary"] = data.groupby(["Department"])["salary"].transform("mean")
```

```python
data.head(5)
```

```python
# Jinki average salary 50000 se kam hai un departments ko hatao (filter).

data = data.groupby(["Department"]).filter(
    lambda x: x["salary"].mean() >= 50000
)
```

```python
data.head(4)
```

```python
# Custom names ke saath agg() use karo.
grp = data.groupby(["Department"]).agg(
    Total_salary = ("salary", "sum"),
    avg_Salary = ("salary", "mean"),
    Max_salary = ("salary", "max")
)
```
