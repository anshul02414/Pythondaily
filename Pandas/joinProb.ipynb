```python
import pandas as pd
```

```python
df1 = pd.read_excel("compney bio.xlsx")
df2 = pd.read_excel("compney salary.xlsx")
```

```python
df1.head(3)
```

```python
df2.head(3)
```

```python
# Employee aur Salary table ko inner join karo.
pd.merge(df1,df2,how="inner")
```

```python
print(pd.merge(df2,df1,how="left"))
```

```python
# Outer join karo.
print(pd.merge(df1,df2,how="outer"))

# also shows if values are not exist
```

```python
# Different column names use karke merge karo.
print(pd.merge(df1,df2,on="eid"))
```

```python
df1.head(5)
```

```python
# Department + City par merge karo.
print(pd.merge(df1,df2,on="Department"))
```

```python
# indicator=True use karo.
print(pd.merge(df1,df2,on="eid", indicator=True))
```
