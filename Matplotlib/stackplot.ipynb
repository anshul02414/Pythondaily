```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
days = [1,2,3,4,5,6,7]

# nop = number of peoples 
nop1 = [88,40,12,45,60,77,90]
nop2 = [29,1,45,8,9,2,77]
nop3 = [88,70,1,45,20,45,22]

plt.stackplot(days, nop1, nop2, nop3)
plt.show()
```
