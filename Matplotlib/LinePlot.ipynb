```python
import matplotlib.pyplot as plt
```

```python
x = ["d1","d2","d3","d4","d5","d6","d7"]
y = [8,9,51,44,26,27,43]
y1 = [10,6,45,30,54,52,2]

plt.plot(x,y, marker="*", ls=":", label="Week1")
plt.plot(x,y1, color="black", alpha=0.4, marker="o", ls="--", label="Week2")

plt.legend()
plt.title("Week costumer Graph")
plt.xlabel("Days")
plt.ylabel("Costumers")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.read_excel("expense3.xlsx")
data.head(5)
```

```python
df = pd.DataFrame(data)
df.head(5)
```

```python
plt.plot(df["Date"], df["Amount"], color="black")
plt.title("Costumer pr day")
plt.xlabel("Date")
plt.ylabel("Amount")

plt.show()
```

```python
plt.plot(df["Category"], df["Amount"])
plt.show()
```

```python
grp = df.groupby("Category")["Amount"].sum()
```

```python
grp
```

```python
plt.plot(grp.index, grp.values, color="black")
plt.title("plot using Grouped by Method")
plt.xlabel("Categoury")
plt.ylabel("Amount")

plt.show()
```
