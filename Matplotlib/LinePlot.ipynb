```python
import matplotlib.pyplot as plt
```

```python
x = ["d1","d2","d3","d4","d5","d6","d7"]
y = [8,9,51,44,26,27,43]
y1 = [10,6,45,30,54,52,2]

plt.plot(x,y, marker="*", ls=":", label="Week1")
plt.plot(x,y1, color="black", alpha=0.4, marker="o", ls="--", label="Week2")

plt.legend()
plt.title("Week costumer Graph")
plt.xlabel("Days")
plt.ylabel("Costumers")
plt.show()
```
