```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import random
```

```python
x = np.random.randint(1,10,50)
y = np.random.randint(10,100,50)

plt.scatter(x,y, marker="D", color="black")
plt.show()
```

```python
color = np.random.randint(1,10,50)

plt.scatter(x,y, marker="o", c=color, cmap="viridis")

plt.title("Scatter plot")
plt.colorbar()
```
