```python
import matplotlib.pyplot as plt
```

```python
brands = ["Oppo", "Samsung", "Vivo", "Apple", "Honor", "Nokia", "Other"]
sales_persent = [12, 26, 18, 25, 7, 3, 8]
ex = [0,0,0,0,0,0, 0.1]

plt.pie(sales_persent, labels=brands, explode=ex, shadow=True, autopct="%.1f")
plt.show()
```
