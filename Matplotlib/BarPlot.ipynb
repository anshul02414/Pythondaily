```python
import matplotlib.pyplot as plt

x = ["A","B","C","D","E"]
y = [22,56,88,24,65]

c = ["black", "red", "blue", "green", "yellow"]

plt.bar(x,y,color=c)
plt.xlabel("Values", size=20)
plt.ylabel("Name", size=18)
plt.show()
```

```python
import pandas as pd
```

```python
data = pd.read_excel("expense3.xlsx")
df = pd.DataFrame(data)

df.head(5)
```

```python
print(df.dtypes)
```

```python
print(df["Payment Mode"].unique())
```

```python
print(df["Payment Mode"].isnull().sum())
```

```python
df["Payment Mode"] = df["Payment Mode"].fillna("Unknown")
```

```python
plt.bar(df["Payment Mode"], df["Amount"], color="black")
plt.xlabel("Payment Method", size=15)
plt.ylabel("Amount in INR", size=15)
plt.show()
```

```python
grp = df.groupby("Payment Mode")["Amount"].sum()
print(grp)
```

```python
plt.bar(grp.index, grp.values)
plt.show
```
